#### Importando Bibliotecas

```python
import pandas as pd
from datetime import datetime, timedelta
```

#### Realizando o Fluxo de Comparação das URLs de Ontem com as de Hoje

```python
# Obter a data de hoje
today = datetime.now().date()

# Obter a data de ontem subtraindo um dia da data de hoje
yesterday = today - timedelta(days=1)

# Formatar as datas para o formato YYYY_MM_DD
today = today.strftime('%Y_%m_%d')
yesterday = yesterday.strftime('%Y_%m_%d')

# Definindo os bairros de busca
bairros = ['santa_cecilia', 'perdizes', 'barra_funda']
```

#### Zap Imóveis:

```python
# Fluxo comparativo para o Zap Imóveis
for bairro in bairros:
    # Leia os arquivos CSV em dois DataFrames separados
    try:
        data_today = pd.read_csv(f'{today}/zapimoveis_{bairro}_{today}.csv')
        data_yesterday = pd.read_csv(f'{yesterday}/zapimoveis_{bairro}_{yesterday}.csv')

        # Compare as URLs presentes em ambos os DataFrames e identifique as URLs únicas
        unique_urls = data_today[~data_today['URL'].isin(data_yesterday['URL'])]['URL'].tolist()
        
        # Salvando as URLs retornadas em um dataframe
        data = pd.DataFrame({'URL': unique_urls})
        
        # Salvando o dataframe em um arquivo .csv
        data.to_csv(f'zapimoveis_{bairro}_novos.csv', index=False)
        
    except Exception as e:
        print("\nOcorreu um erro:", e)
```

```python
# Fluxo comparativo para o Quinto Andar
for bairro in bairros:
    # Leia os arquivos CSV em dois DataFrames separados
    try:
        data_today = pd.read_csv(f'{today}/quintoandar_{bairro}_{today}.csv')
        data_yesterday = pd.read_csv(f'{yesterday}/quintoandar_{bairro}_{yesterday}.csv')

        # Compare as URLs presentes em ambos os DataFrames e identifique as URLs únicas
        unique_urls = data_today[~data_today['URL'].isin(data_yesterday['URL'])]['URL'].tolist()
        
        # Salvando as URLs retornadas em um dataframe
        data = pd.DataFrame({'URL': unique_urls})
        
        # Salvando o dataframe em um arquivo .csv
        data.to_csv(f'quintoandar_{bairro}_novos.csv', index=False)
        
    except Exception as e:
        print("\nOcorreu um erro:", e)
```
