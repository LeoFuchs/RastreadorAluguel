#### Importando Bibliotecas

```python
import requests
import time
import os
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
from datetime import datetime
```

#### Função Auxiliar para Extração dos Dados

```python
def extrair_codigo_fonte_com_rolagem(url):
    try:
        # Inicializa o driver do Chrome
        driver = webdriver.Chrome()
        
        # Abre a URL no navegador
        driver.get(url)
        
        # Aguarda um tempo para a página carregar
        time.sleep(5)
        
        # Inicializa a variável para armazenar o código fonte
        codigo_fonte = ""

        # Função para verificar se o botão "Ver mais" está visível
        def is_button_visible():
            return len(driver.find_elements(By.XPATH, "//button[contains(text(), 'Ver mais')]")) > 0
            
        # Loop para clicar no botão "Ver mais" enquanto estiver visível
        while is_button_visible():
            # Clica no botão "Ver mais"
            driver.find_element(By.XPATH, "//button[contains(text(), 'Ver mais')]").click()
    
            # Aguarda alguns segundos para carregar os novos resultados (ajuste conforme necessário)
            time.sleep(7)

        # Após todos os resultados serem carregados, obtém o HTML da página atual
        codigo_fonte = driver.page_source
        
        # Fecha o navegador
        driver.quit()
        
        return codigo_fonte
    except Exception as e:
        print("Ocorreu um erro:", e)
        return None
```

#### 1. Extração dos Dados (Santa Cecília)

Queremos extrair os dados de uma busca fixa, com os seguintes parâmetros:

- Bairro: Santa Cecília
- Cidade: São Paulo - SP
- Tipo: Apartamento residencial, Studio residencial, Kitnet residencial, Flat residencial e Loft residencial 
- Preço Total Mínimo: R\$ 3000,00
- Preço Total Máximo: R\$ 6000,00
- Área Mínima: 45m²
- Área Máxima: 110m²

que geram a seguinte URL: 

**https://www.quintoandar.com.br/alugar/imovel/santa-cecilia-sao-paulo-sp-brasil/de-3000-a-6000-reais/apartamento/kitnet/de-45-a-110-m2**

```python
url_busca = f"https://www.quintoandar.com.br/alugar/imovel/santa-cecilia-sao-paulo-sp-brasil/de-3000-a-6000-reais/apartamento/kitnet/de-45-a-110-m2"

codigo_fonte = extrair_codigo_fonte_com_rolagem(url_busca)
print(f"Código fonte da página obtido com sucesso!")
```

```python
# Inicializar uma lista para armazenar as URLs
urls = []

soup = BeautifulSoup(codigo_fonte, 'lxml')

# Encontrar todos os elementos <div> com o atributo data-position
divs_com_data_position = soup.find_all('div', {'data-testid': 'house-card-container-rent'})
    
# Iterar sobre os elementos <div> encontrados
for div in divs_com_data_position:
    # Encontrar todos os elementos <a> dentro do <div>
    links = div.find_all('a', href=True)
    # Extrair as URLs e adicioná-las à lista de URLs
    for link in links:
        urls.append(link['href'])
print(f"URLs da página salvo com sucesso!") 
```

```python
# Especificando o bairro da busca para salvamento dos dados
bairro = 'santa_cecilia'

# Expondo o tamanho do arquivo de hoje
print(f"Hoje tivemos {len(urls)} URLs!")

# Salvando as URLs retornadas em um dataframe
data = pd.DataFrame({'URL': urls})

# Obtenha a data de hoje
today = datetime.now().strftime('%Y_%m_%d')

# Criar o caminho completo do arquivo
caminho_arquivo = f"{today}/quintoandar_{bairro}_{today}.csv"

# Criar a pasta se ela não existir
if not os.path.exists(today):
    os.makedirs(today)

# Criar o caminho completo do arquivo
caminho_arquivo = f"{today}/quintoandar_{bairro}_{today}.csv"

# Criar a pasta se ela não existir
if not os.path.exists(today):
    os.makedirs(today)

# Salvando o dataframe em um arquivo .csv com o formato 'quintoandar_bairro_YYYY_MM_DD.csv'
data.to_csv(caminho_arquivo, index=False)
```
